## Q1 The idea of statistical learning

### (a) In the model $Y = f(X) + \varepsilon$, explain what $Y$, $X$, $f$ and $\varepsilon$ are, using the Advertising data (sales; TV, radio and newspaper budgets) as the example.

In the statistical learning model $Y = f(X) + \varepsilon$, each term corresponds to a concrete component of the Advertising dataset:
* **$Y$ (Response / Dependent Variable):** This is the quantitative target variable we wish to predict or explain, which represents product **`sales`** (measured in thousands of units).
* **$X$ (Predictors / Independent Variables / Features):** This is the input vector $X = (X_1, X_2, X_3)$ denoting the advertising budgets allocated across the three media channels: **`TV`**, **`radio`**, and **`newspaper`** (each measured in thousands of dollars).
* **$f$ (Systematic Function):** This represents the true, unknown underlying mathematical function that characterizes the systematic relationship between the advertising expenditures $X$ and the expected sales $Y$.
* **$\varepsilon$ (Random Error Term):** This is a random error component with mean zero that is independent of $X$. It captures unobserved factors influencing sales that are omitted from the model (such as product quality, branding, competitor actions, or macroeconomic conditions) as well as inherent measurement noise.

### (b) Why can we not predict sales perfectly, even if we knew the true $f$? Use the words reducible and irreducible error.

Even if we knew the exact, true systematic relationship $f$, we could never predict sales with perfect accuracy due to the fundamental difference between **reducible error** and **irreducible error**. The total expected prediction error decomposes as $\mathrm{E}[(Y - \hat{Y})^2] = [f(X) - \hat{f}(X)]^2 + \mathrm{Var}(\varepsilon)$. The first term, $[f(X) - \hat{f}(X)]^2$, represents the **reducible error**, which stems from inaccuracies in estimating $f$ and can be minimized by selecting better learning algorithms or collecting more data. In contrast, the second term, $\mathrm{Var}(\varepsilon)$, represents the **irreducible error**, which arises from inherent real-world randomness and unmeasured variables contained within $\varepsilon$. Because $\mathrm{Var}(\varepsilon) > 0$, irreducible error provides an inevitable lower bound on prediction error that no model can eliminate, preventing sales from being predicted perfectly.

## Q3 Auto data by region

In the `Auto` dataset, the column `origin` is encoded as:
* `1` = USA
* `2` = Europe
* `3` = Japan

```python
# Q3(a): How many cars come from each origin? (Using value_counts())
origin_counts = Auto['origin'].value_counts().sort_index()

origin_labels = {1: 'USA', 2: 'Europe', 3: 'Japan'}
origin_table = pd.DataFrame({
    'Origin Code': origin_counts.index,
    'Region': [origin_labels[i] for i in origin_counts.index],
    'Car Count': origin_counts.values
})

print("Q3(a) Number of cars by region:")
display(origin_table)
```

```python
# Q3(b): Compute the average mpg for each origin with groupby
avg_mpg = Auto.groupby('origin')['mpg'].mean()

avg_mpg_table = pd.DataFrame({
    'Origin Code': avg_mpg.index,
    'Region': [origin_labels[i] for i in avg_mpg.index],
    'Average MPG': avg_mpg.values
})

print("Q3(b) Average mpg by region:")
display(avg_mpg_table.round(2))
```

```python
# Q3(c): Make a boxplot of mpg by origin
plt.figure(figsize=(8, 5.5))

mpg_grouped = [Auto[Auto['origin'] == o]['mpg'] for o in [1, 2, 3]]

box = plt.boxplot(
    mpg_grouped,
    patch_artist=True,
    medianprops=dict(color='red', linewidth=2),
    boxprops=dict(facecolor='#d0e1fd', color='#084298', linewidth=1.2),
    whiskerprops=dict(color='#084298', linewidth=1.2),
    capprops=dict(color='#084298', linewidth=1.2),
    flierprops=dict(marker='o', color='darkblue', alpha=0.6)
)

plt.xticks([1, 2, 3], ['USA (1)', 'Europe (2)', 'Japan (3)'])

plt.title('Fuel Efficiency (mpg) by Region of Origin', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Region of Origin', fontsize=12, labelpad=8)
plt.ylabel('Miles Per Gallon (mpg)', fontsize=12, labelpad=8)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Q3(d) Region comparison and explanation
Japan (Origin 3) produces the most fuel-efficient cars, achieving the highest average fuel economy at approximately $30.45$ mpg compared to $27.60$ mpg for Europe and $20.03$ mpg for the USA. A plausible reason is that Japanese car manufacturers during the 1970s and 1980s prioritized compact, lightweight vehicle designs powered by efficient four-cylinder engines in response to domestic resource constraints and the global oil crises, whereas American manufacturers traditionally produced much heavier cars with large-displacement V8 engines.

## Q4 TV or radio?

Budgets are in thousands of dollars; sales are in thousands of units.

```python
# Q4(a): Fit straight line of sales on TV and on radio with np.polyfit, and compute correlation
# np.polyfit(x, y, deg=1) returns [b1 (slope), b0 (intercept)]

# TV linear fit
tv_b1, tv_b0 = np.polyfit(ad['TV'], ad['sales'], deg=1)
tv_corr = ad['TV'].corr(ad['sales'])

# Radio linear fit
radio_b1, radio_b0 = np.polyfit(ad['radio'], ad['sales'], deg=1)
radio_corr = ad['radio'].corr(ad['sales'])

# Summary table reporting b0, b1 and correlation
q4_table = pd.DataFrame({
    'Channel': ['TV', 'Radio'],
    'b0 (Intercept)': [tv_b0, radio_b0],
    'b1 (Slope)': [tv_b1, radio_b1],
    'Correlation with Sales (r)': [tv_corr, radio_corr]
})

print("Q4(a) Fitted parameters and correlations:")
display(q4_table.round(4))
```

### Q4(b) Explanation of the slope $b_1$ of the radio line
The slope of the radio regression line is $b_1 \approx 0.2025$. This means that for each additional \$1,000 spent on the radio advertising budget (as budgets are in thousands of dollars), product sales are expected to increase on average by approximately $0.2025$ thousand units, which corresponds to roughly $202.5$ additional units sold.

```python
# Q4(c): Predict sales when TV = 150, using the TV line
tv_val = 150
pred_sales_150 = tv_b0 + tv_b1 * tv_val

print(f"TV Fitted Equation: Sales = b0 + b1 * TV")
print(f"Values: b0 = {tv_b0:.4f}, b1 = {tv_b1:.4f}")
print(f"Calculation: Predicted Sales = {tv_b0:.4f} + {tv_b1:.4f} * {tv_val}")
print(f"Predicted Sales = {pred_sales_150:.4f} thousand units ({pred_sales_150 * 1000:.1f} units)")
```

### Q4(d) Comparison of TV and radio
TV advertising has the stronger linear relationship with sales, as reflected by its substantially higher correlation coefficient ($r \approx 0.7822$ for TV versus $r \approx 0.5762$ for radio). However, radio advertising has the larger slope ($b_1 \approx 0.2025$ for radio versus $b_1 \approx 0.0475$ for TV), indicating a higher expected unit increase in sales per dollar spent. It is not fair or statistically sound to compare the two slopes directly because the two media channels have vastly different budget scales and variances (TV spending has a much wider spread, $\mathrm{SD} \approx 85.85$, compared to radio, $\mathrm{SD} \approx 14.85$), and separate univariate models ignore the synergy (interaction effect) that exists when multiple channels are used together.
